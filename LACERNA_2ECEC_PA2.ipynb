# Problem #1: Reproducible Normalization

```python
#Important Convention rule
import numpy as np

np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))

# Normalizing array using the given formula
X_normalized = (X - X.mean()) / np.std(X)

# Display all required checks
print("The randomized array is:\n", X)
print("The normalized array is:\n", X_normalized)
print("The mean of the normalized array is:", X_normalized.mean())
print("The standard deviation of the normalized array is:", np.std(X_normalized))

# Save normalized array
np.save("X_normalized.npy", X_normalized)
```

## Problem #2: Cubes Divisible by 4 problem

```python
import numpy as np

# Creating the first 100 positive integers and using a boolean condition
C = np.arange(1, 101).reshape(10, 10) ** 3
div_by_4 = C[C % 4 == 0]

# Display all required checks
print("The shape of the original array is:", C.shape)
print("The array of the elements divisible by 4 is:\n", div_by_4)
print("The number of elements that are divisible by 4 is:", div_by_4.size)

# Saving array
np.save("div_by_4.npy", div_by_4)
```

## Problem #3 : Above-mean squares

```python
import numpy as np
# PROBLEM C: Above-Mean Squares

# creating 6x6 array (S)
S = (np.arange(1, 37) ** 2).reshape(6, 6)
# compute mean of all elements of S and store to S_mean
S_mean = S.mean()  
# stored values into above_mean
above_mean = S[S > S_mean]

# Display required checks
print("The original array is:\n", S)
print("The mean of the array is:", S_mean)
print("The elements that are greater than the mean are:\n", above_mean)
print("The number of selected elements is:", above_mean.size)  # <-- Missing check

# Saving  array
np.save("above_mean.npy", above_mean)
```
